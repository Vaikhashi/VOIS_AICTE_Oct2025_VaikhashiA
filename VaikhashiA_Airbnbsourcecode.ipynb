```python
import pandas as pd
df = pd.read_csv('/content/1730285881-Airbnb_Open_Data.csv')
df.head()
```

```python
df.info()
df.describe(include='all')
```

```python
df.isnull().sum().sort_values(ascending=False).head(15)
```

```python

df['price'] = df['price'].astype(str).str.replace(r'[\$,]', '', regex=True)
df['price'] = pd.to_numeric(df['price'], errors='coerce')
df['price'].head()
```

```python
df.columns
```

```python
df['room type'].value_counts().plot(kind='bar', color='skyblue')
plt.title('Number of Listings by Room Type')
plt.xlabel('Room Type')
plt.ylabel('Count')
plt.show()
```

```python
df['neighbourhood group'].value_counts().plot(kind='bar', color='lightgreen')
plt.title('Listings by Neighbourhood Group')
plt.xlabel('neighbourhood group')
plt.ylabel('Count')
plt.show()
```

```python
avg_price = df.groupby('neighbourhood')['price'].mean().sort_values(ascending=False).head(10)
avg_price.plot(kind='bar', color='coral')
plt.title('Top 10 Neighbourhoods by Average Price')
plt.ylabel('Average Price')
plt.show()
```

```python
sns.boxplot(x='room type', y='price', data=df)
plt.title('Room Type vs Price')
plt.show()
```

```python
top_hosts = df['host id'].value_counts().head(10)
top_hosts.plot(kind='bar', color='orchid')
plt.title('Top 10 Hosts by Number of Listings')
plt.xlabel('host id')
plt.ylabel('Number of Listings')
plt.show()
```

```python
if 'host_identity_verified' in df.columns and 'review rate number' in df.columns:
    verified_avg = df.groupby('host_identity_verified')['review rate number'].mean()
    verified_avg.plot(kind='bar', color='gold')
    plt.title('Average Review Rating: Verified vs Not Verified Hosts')
    plt.ylabel('Average Rating')
    plt.show()
else:
    print("Required columns not found — skipping this question.")
```

```python
if 'service_fee' in df.columns:
    df['service_fee'] = df['service_fee'].astype(str).str.replace(r'[\$,]', '', regex=True)
    df['service_fee'] = pd.to_numeric(df['service_fee'], errors='coerce')
    sns.scatterplot(x='price', y='service_fee', data=df, alpha=0.5)

  plt.title('Price vs Service Fee')
    plt.show()
    print("Correlation:", df['price'].corr(df['service_fee']))
else:
    print("No service_fee column found.")
```
